# A first model

> Needs: [Exploring the listings](00a_exploring_the_listings.ipynb), whose decisions this notebook applies; linear regression, decision trees and the train/test split.

A first model prices each home from the few columns that obviously matter to a price: how big the home is, and how far it is from the city centre. It is fitted with the simplest workflow there is (a random split, a linear regression and a decision tree), and its error is the number that every later model has to beat.

## The listings and the columns

The decisions of [Exploring the listings](00a_exploring_the_listings.ipynb), each explained there: Madrid's listings, without the one 415 km from the centre; the size of each home and its distance to the centre as the features; the asking price as the target.

```
from pathlib import Path

import pandas as pd

DATA_DIR = Path("data")

listings = pd.read_parquet(DATA_DIR / "idealista18_listings.parquet")
listings.shape
```

```
FEATURES = ["built_area_m2", "bedrooms", "bathrooms", "dist_city_centre_km"]
TARGET = "price_eur"

madrid = listings[(listings["city"] == "madrid") & (listings["dist_city_centre_km"] <= 50)]
len(madrid)
```

## Features and target

The features, `X`, and the target, `y`, as scikit-learn names them:

```
X = madrid[FEATURES]
y = madrid[TARGET]
X.shape, y.shape
```
